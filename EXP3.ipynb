```python
#first and last interchange
string = input("Enter a string: ")
new_string = string[-1]+string[1:-1]+string[0]
print("New string: ",new_string)
```

```python
#string made of first 2 and last 2 chars
string = input("Enter a string: ")
new_string = string[:2] + string[-2:]
print("New String: ",new_string)
```

```python
#single string seperated with space from two strings by swapping the char at position 1
s1 = input("Enter First string: ")
s2 = input("Enter Second String: ")
new_s1 = s2[0] + s1[1:]
new_s2 = s1[0] + s2[1:]
print(new_s1 +" "+ new_s2)
```
