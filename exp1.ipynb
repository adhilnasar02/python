```python
#integer
n=input("Enter a number: ")
nn=n+n
nnn=n+n+n
sum=int(n)+int(nn)+int(nnn)
print(n,"+",nn,"+",nnn,"=",sum)
```

```python
#area of a cicle 
import math
rad=int(input("Enter the Radius"))
area=math.pi*rad*rad
print("Area of circle",round(area, 2))
```
