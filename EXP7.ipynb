```python
#colors that contain in set1 not in set2
set_1 = {"Black","Green","Blue","Red","Yellow"}
set_2 = {"Green","Violet","Silver","Gold","Red"}
difference = set_1-set_2
print("Colors in set-1 not in set-2: ",difference)
```

```python
#all colors from set1 and set2
set_1 = {"Black","Green","Blue","Red","Yellow"}
set_2 = {"Green","Violet","Silver","Gold","Red"}
print("colors from set-1 and set-2: ")
print(set_1.union(set_2))
```
