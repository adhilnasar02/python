```python
# Colors that are in set1 but not in set2
set_1 = set(input("Enter colors for set-1 (comma-separated): ").split(","))
set_2 = set(input("Enter colors for set-2 (comma-separated): ").split(","))
difference = set_1 - set_2
print("Colors in set-1 not in set-2:", difference)
```

```python
# All colors from set1 and set2
set_1 = set(input("Enter colors for set-1 (comma-separated): ").split(","))
set_2 = set(input("Enter colors for set-2 (comma-separated): ").split(","))
print("Colors from set-1 and set-2:")
print(set_1.union(set_2))
```
