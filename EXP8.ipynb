```python
# Accept 5 students' names and roll numbers and display in ascending and descending
roll1 = int(input("Enter roll number of student 1: "))
name1 = input("Enter name of student 1: ")
roll2 = int(input("Enter roll number of student 2: "))
name2 = input("Enter name of student 2: ")
roll3 = int(input("Enter roll number of student 3: "))
name3 = input("Enter name of student 3: ")
roll4 = int(input("Enter roll number of student 4: "))
name4 = input("Enter name of student 4: ")
roll5 = int(input("Enter roll number of student 5: "))
name5 = input("Enter name of student 5: ")
students = {
    roll1: name1,
    roll2: name2,
    roll3: name3,
    roll4: name4,
    roll5: name5
}
print("Original dictionary:")
print(students)
ascending = dict(sorted(students.items()))
descending = dict(sorted(students.items(), reverse=True))
print("Ascending order:")
print(ascending)
print("Descending order:")
print(descending)
```

```python
#Merging dictionaries 1 and 2
roll1 = int(input("Enter roll number of student 1: "))
name1 = input("Enter name of student 1: ")
roll2 = int(input("Enter roll number of student 2: "))
name2 = input("Enter name of student 2: ")
roll3 = int(input("Enter roll number of student 3: "))
name3 = input("Enter name of student 3: ")
dic1 = {
    roll1: name1,
    roll2: name2,
    roll3: name3
}
roll4 = int(input("Enter roll number of student 4: "))
name4 = input("Enter name of student 4: ")
roll5 = int(input("Enter roll number of student 5: "))
name5 = input("Enter name of student 5: ")
dic2 = {
    roll4: name4,
    roll5: name5
}
print("First dictionary:")
print(dic1)
print("Second dictionary:")
print(dic2)
dic1.update(dic2)
print("Merged dictionary:")
print(dic1)
```
