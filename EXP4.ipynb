```python
#Display first and last in the list
colors = input("Enter Coluors: ")
color_list = colors.split(",")
print("First color:",color_list[0])
print("Last color:",color_list[-1])
```

```python
#accept filename from user and print extension of it
file=input("Enter Filename: ")
extension = file.split(".")[-1]
print("Extension: ",extension)
```

```python
#accept numbers from user and generate list and it'sum
numbers = input("Enter Numbers(comma-seperated): ")
list = [int (x)for x in numbers.split(",")]
print("List: ",list)
print("Sum: ",sum(list))
```
